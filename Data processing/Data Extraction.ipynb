5.Extracting Data from APIs(استخراج داده از APIها)
To extract data from APIs, you can use libraries like requests in Python
برای استخراج داده از ای پی ای، می‌توانید از کتابخانه‌هایی مانند اجازه در پایتون استفاده کنید.

```python
# Extracting Data from APIs(استخراج داده از APIها)
import requests

response = requests.get('https://api.example.com/data')
data = response.json()
print(data)
```

6.Extracting Data from Web Pages(استخراج داده از صفحات وب )
To extract data from web pages, you can use libraries like BeautifulSoup and requests in Python
برای استخراج داده از صفحات وب، می‌توانید از کتابخانه‌هایی مانند BeautifulSoup و requests در پایتون استفاده کنید.

```python
# Extracting Data from Web Pages(استخراج داده از صفحات وب )
from bs4 import BeautifulSoup
import requests

url = 'https://example.com'
response = requests.get(url)
soup = BeautifulSoup(response.text, 'html.parser')
print(soup.title.text)
```

7.Cleaning and Preparing Data(پاکسازی و آماده‌سازی داده )
After extracting data, you may need to clean and prepare the data. This step includes removing duplicates, filling missing values, and converting formats
 پس از استخراج داده، ممکن است نیاز به پاکسازی و آماده‌سازی داده‌ها داشته باشید. این مرحله شامل حذف داده‌های تکراری، پر کردن مقادیر خالی و تبدیل فرمت‌ها است.

```python
# Cleaning and Preparing Data(پاکسازی و آماده‌سازی داده )
data.drop_duplicates(inplace=True)
data.fillna(0, inplace=True)
```

8.Storing Extracted Data(ذخیره‌سازی داده‌های استخراج شده)
After extracting and cleaning the data, you can store it in various formats such as CSV, databases, or JSON files
پس از استخراج و پاکسازی داده‌ها، می‌توانید آن‌ها را در قالب‌های مختلف مانند CSV، پایگاه داده یا فایل‌های JSON ذخیره کنید.

```python
# Storing Extracted Data(ذخیره‌سازی داده‌های استخراج شده)
data.to_csv('cleaned_data.csv', index=False)
```
