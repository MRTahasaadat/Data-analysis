```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

# Creating a sample DataFrame( ایجاد یک دیتافریم نمونه)
data = {'Feature1': [10, 20, 30, 40, 50],
        'Feature2': [100, 200, 300, 400, 500]}
df = pd.DataFrame(data)

# Creating a MinMaxScaler object( ابداع یک شیء MinMaxScaler)
scaler = MinMaxScaler()

# Scaling the data(مقیاس‌بندی دادها)
scaled_data = scaler.fit_transform(df)

# Converting scaled data back to a DataFrame(تبدیل داده‌های مقیاس‌بندی شده به دیتافریم)
scaled_df = pd.DataFrame(scaled_data, columns=df.columns)

print("Original Data:")
print(df)
print("\nScaled Data using Min-Max Scaling:")
print(scaled_df)
```

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler

# Creating a sample DataFrame( ایجاد یک دیتافریم نمونه)
data = {'Feature1': [10, 20, 30, 40, 50],
        'Feature2': [100, 200, 300, 400, 500]}
df = pd.DataFrame(data)

# Creating a StandardScaler object( ابداع یک شیء StandardScaler)
scaler = StandardScaler()

# Scaling the data( ابداع یک شیء StandardScaler)
scaled_data = scaler.fit_transform(df)

# Converting scaled data back to a DataFrame( تبدیل داده‌های مقیاس‌بندی شده به دیتافریم)
scaled_df = pd.DataFrame(scaled_data, columns=df.columns)

print("Original Data:")
print(df)
print("\nScaled Data using Standardization:")
print(scaled_df)
```
