پاکسازی داده / Data Cleaning 


Data cleaning is the process of preparing data for analysis. Its goal is to remove or correct data that is incorrect, incomplete, duplicated, or irrelevant. A clean dataset is essential to ensure the accuracy and reliability of analysis results.

پاکسازی داده فرآیند آماده‌سازی داده‌ها برای تحلیل است. هدف آن حذف یا اصلاح داده‌های نادرست، ناقص، تکراری یا نامربوط است. یک مجموعه داده تمیز برای اطمینان از دقت و قابلیت اطمینان نتایج تحلیل ضروری است.

Main Steps of Data Cleaning:

مراحل اصلی پاکسازی داده:

1. in Steps of Data Cleaning:

    Data Inspection:
        Overview of the data to identify potential problems such as missing values, outliers, and inconsistencies.
        Using functions like head(), info(), describe() in Pandas.

1.بررسی داده‌ها (Data Inspection):

    بررسی اجمالی داده‌ها برای شناسایی مشکلات احتمالی مانند مقادیر از دست رفته، ناهنجاری‌ها و ناسازگاری‌ها.
    استفاده از توابع مانند head()‎، info()‎، describe()‎ در Pandas.



```python
import pandas as pd

# Reading CSV file(خواندن فایل CSV)
df = pd.read_csv('data.csv')

# Displaying the first few rows(نمایش چند سطر اول)
print(df.head())

# General information about the data(اطلاعات کلی در مورد داده‌ها)
print(df.info())

# Descriptive statistics of the data( آمار توصیفی داده‌ها)
print(df.describe())
```

2.Handling Missing Values(مدیریت مقادیر از دست رفته):

Identifying: Identifying columns and rows that have missing values.
Imputation: Filling missing values with appropriate values (such as mean, median, mode, or a fixed value).
Removal: Removing rows or columns with a large number of missing values.

شناسایی: شناسایی ستون‌ها و سطرهایی که مقادیر از دست رفته دارند.
جایگزینی (Imputation): پر کردن مقادیر از دست رفته با مقادیر مناسب (مانند میانگین، میانه، مد، یا یک مقدار ثابت).
حذف: حذف سطرها یا ستون‌هایی با تعداد زیادی مقدار از دست رفته.

```python
# Checking missing values( بررسی مقادیر از دست رفته)
print(df.isnull().sum())

# Replacing missing values with the mean( جایگزینی مقادیر از دست رفته با میانگین)
df.fillna(df.mean(), inplace=True)

# Removing rows with missing values( حذف سطرها با مقادیر از دست رفته)
df.dropna(inplace=True)
```

3.Removing Duplicates(حذف داده‌های تکراری):

Identifying: Identifying duplicate rows in the data.
Removal: Removing duplicate rows to prevent inappropriate influence on analysis.
شناسایی: شناسایی سطرهای تکراری در داده‌ها.
حذف: حذف سطرهای تکراری برای جلوگیری از تأثیر نامناسب بر تحلیل.

```python
# Identifying duplicate rows(شناسایی سطرهای تکراری)
print(df.duplicated().sum())

# Removing duplicate rows( حذف سطرهای تکراری)
df.drop_duplicates(inplace=True)
```

4. Handling Outliers(اصلاح ناهنجاری‌ها):
Identifying: Identifying data that differs significantly from other data.
Removal or Replacement: Removing outliers or replacing them with more appropriate values.
شناسایی: شناسایی داده‌هایی که به طور قابل توجهی از سایر داده‌ها متفاوت هستند.
حذف یا جایگزینی: حذف ناهنجاری‌ها یا جایگزینی آنها با مقادیر مناسب‌تر.

```python
import numpy as np
from scipy import stats

# Removing outliers based on Z-score( حذف ناهنجاری‌ها بر اساس Z-score)
z = np.abs(stats.zscore(df['column_name']))
threshold = 3
df = df[z < threshold]
```

5.Data Standardization and Normalization(استانداردسازی و نرمال‌سازی داده):
Standardization: Scaling data so that the mean is 0 and the standard deviation is 1.
Normalization: Scaling data to a specific range (e.g., 0 to 1).

استانداردسازی: تغییر مقیاس داده‌ها به طوری که میانگین 0 و انحراف معیار 1 داشته باشند.
نرمال‌سازی: تغییر مقیاس داده‌ها به یک محدوده مشخص (مثلاً 0 تا 1).

```python
from sklearn.preprocessing import StandardScaler, MinMaxScaler

#Standardization(استانداردسازی)
scaler = StandardScaler()
df['column_name'] = scaler.fit_transform(df[['column_name']])

#Normalization(نرمال‌سازی)
minmax_scaler = MinMaxScaler()
df['column_name'] = minmax_scaler.fit_transform(df[['column_name']])
```

6.Data Type Conversion(تبدیل نوع داده‌ها):

Changing the data type of columns to appropriate types (such as converting string to date).
تغییر نوع داده‌های ستون‌ها به انواع مناسب (مانند تبدیل رشته به تاریخ).

```python
## Converting a column to date type(تبدیل ستون به نوع تاریخ)
df['date_column'] = pd.to_datetime(df['date_column'])

## Converting a column to numeric type(تبدیل ستون به نوع عددی)
df['numeric_column'] = df['numeric_column'].astype(float)
```

7.Handling Inconsistencies(اصلاح ناسازگاری‌ها):

Standardizing formats, correcting spelling errors, and fixing other inconsistencies.
یکسان‌سازی فرمت‌ها، تصحیح غلط‌های املایی و اصلاح سایر ناسازگاری‌ها.


```python
# Replacing values(جایگزینی مقادیر)
df['column_name'] = df['column_name'].replace({'old_value': 'new_value'})

# Converting text to lowercase(تبدیل متن به حروف کوچک)
df['column_name'] = df['column_name'].str.lower()
```
