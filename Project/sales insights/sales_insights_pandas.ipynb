```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = {
    "date":["2025-01-01","2025-01-02","2025-01-03","2025-01-04","2025-01-05"],
    "product" : ["Laptop","phone", "tablet", "airpods","hedphone"],
    "Quantity" : [1,2,3,1,2],
    "Price" : [1000, 1000, 500, 700,500]
}

df = pd.DataFrame(data)
df["date"] = pd.to_datetime(df["date"])
df.head()
```

```python
df.describe()
```

```python
df["Revenue"] = df["Quantity"] * df["Price"]
df.groupby("product")["Revenue"].sum()
```

```python
df.groupby("product")["Quantity"].sum().sort_values(ascending=False)
```

```python
sns.barplot(x = "product", y = "Quantity", data = df)
plt.title("Top selling Products")
plt.show()
```

```python
daily_sales = df.groupby("date")["Revenue"].sum().reset_index()
sns.lineplot(x="date", y= "Revenue",data = daily_sales , marker = "o")
plt.title("Revenue Over Time")
plt.show()
```
