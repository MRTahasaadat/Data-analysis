```python
import matplotlib.pyplot as plt
import numpy as np
```

Types of charts in Matplotlib

```python
# Line Plot(نمودار خطی)
# data(داده)
x = np.linspace(0, 10, 100) #Create 100 points between 0 and 10
y = np.sin(x)

# Create a chart(ایجاد نمودار)
plt.figure(figsize=(8, 6))  # Determine the size of the chart(تعیین اندازه نمودار)
plt.plot(x, y, label='sin(x)', color='blue', linestyle='-')

# Create a chart(افزودن عنوان و برچسب‌ها)
plt.title('Sine linear graph')
plt.xlabel('x')
plt.ylabel('sin(x)')

# Add guide(افزودن راهنما)
plt.legend()

#Show chart(نمایش نمودار)
plt.grid(True)  # Add grid lines(اضافه کردن خطوط شبکه)
plt.show()
```

```python
# Scatter Plot(نمودار پراکندگی)
# data(داده)
x = np.random.rand(50)  
y = np.random.rand(50)
colors = np.random.rand(50)  
sizes = 100 * np.random.rand(50)  

# Create a chart(ایجاد نمودار)
plt.figure(figsize=(8, 6))
plt.scatter(x, y, c=colors, s=sizes, alpha=0.5, cmap='viridis')

# Create a chart(افزودن عنوان و برچسب‌ها)
plt.title('Scatter Plot')
plt.xlabel('x')
plt.ylabel('y')

# Add a color bar(اضافه کردن نوار رنگ)
plt.colorbar(label='color')

#Show chart(نمایش نمودار)
plt.show()
```

```python
# Bar Chart(نمودار میله‌ای )

# data(داده)
categories = ['A', 'B', 'C', 'D', 'E']
values = np.random.randint(10, 100, len(categories))  

# Create a chart(ایجاد نمودار)
plt.figure(figsize=(8, 6))
plt.bar(categories, values, color='skyblue')

# Create a chart(افزودن عنوان و برچسب‌ها)
plt.title('Bar chart')
plt.xlabel('group')
plt.ylabel('Values')

#Show chart(نمایش نمودار)
plt.show()
```

```python
# Histogram(هیستوگرام)

# data(داده)
data = np.random.randn(1000)  

# Create a chart(ایجاد نمودار)
plt.figure(figsize=(8, 6))
plt.hist(data, bins=30, color='lightgreen', edgecolor='black')

# Create a chart(افزودن عنوان و برچسب‌ها)
plt.title('Histogram')
plt.xlabel('Values')
plt.ylabel('Abundance')

#Show chart(نمایش نمودار)
plt.show()
```

```python
# Pie Chart(نمودار دایره‌ای)

# data(داده)
data = np.random.randn(100, 5)  

# Create a chart(ایجاد نمودار)
plt.figure(figsize=(8, 6))
plt.boxplot(data, labels=['A', 'B', 'C', 'D', 'E'], patch_artist=True)

# Create a chart(افزودن عنوان و برچسب‌ها)
plt.title('Box plot')
plt.xlabel('group')
plt.ylabel('Values')

#Show chart(نمایش نمودار)
plt.show()
```

Plot Elements

```python
# color => Color of line or dots(رنگ خط یا نقاط)
# linestyle => Line type(نوع خط)
# linewidth => Line thickness(ضخامت خط)
# marker => Pointer type for points(نوع نشانگر برای نقاط )
# markersize => Marker size(اندازه نشانگرها)
# label => Label for line or points used in the legend(برچسب برای خط یا نقاط که در راهنما (legend) استفاده می‌شود)

x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.plot(x, y, color='blue', linestyle='-', linewidth=2, marker='o', markersize=5, label='sin(x)')
plt.legend()
plt.show()
```

scatter Elements

```python
# s => The size of the points (can be a constant value or an array).(اندازه نقاط (می‌تواند یک مقدار ثابت یا یک آرایه باشد))
# c => The color of the dots (can be a constant value or an array)(رنگ نقاط (می‌تواند یک مقدار ثابت یا یک آرایه باشد))
# marker => Pointer type for points(نوع نشانگر برای نقاط)
# alpha => Transparency of the points (between 0 , 1).(شفافیت نقاط (بین 0 و 1))
# cmap => Color map to color points based on another value(نقشه رنگی برای رنگ‌بندی نقاط بر اساس یک مقدار دیگر)
# label => Labels for points used in the guide(برچسب برای نقاط که در راهنما استفاده می‌شود)

x = np.random.rand(50)
y = np.random.rand(50)
colors = np.random.rand(50)
sizes = 100 * np.random.rand(50)

plt.scatter(x, y, s=sizes, c=colors, marker='o', alpha=0.5, label='Random Points')
plt.legend()
plt.show()
```

bar,barh Elements

```python
# x => A list of x values ​​(positions of the bars)(آرایه‌ای از  x (موقعیت میله‌ها))
# height => An array of bar heights(آرایه‌ای از ارتفاع میله‌ها)
# width => Bar width (default 0.8)(عرض میله‌ها (پیش‌فرض 0.8))
# color => Color of the bars(رنگ میله‌ها)
# alpha =>  => The transparency of the bars(شفافیت میله‌ها)
# label => Label for bars used in the guide(برچسب برای میله‌ها که در راهنما استفاده می‌شود)
# align =>  => Alignment of bars relative to x position (default 'center')(تراز میله‌ها نسبت به موقعیت x (پیش‌فرض 'center'))
# edgecolor => Bar edge color(رنگ لبه میله‌ها)

categories = ['A', 'B', 'C', 'D']
values = np.random.randint(10, 100, 4)

plt.bar(categories, values, color='green', alpha=0.7, label='Values')
plt.legend()
plt.show()
```

```python
categories = ['A', 'B', 'C', 'D']
values = np.random.randint(10, 100, 4)

plt.barh(categories, values, color='green', alpha=0.7, label='Values')
plt.legend()
plt.show()
```

hist Elements

```python
# x => An array of data (آرایه‌ای از داده‌ها)
# bins => Number of bins or bin edges => (تعداد بازه‌ها (bins) یا لبه‌های بازه‌ها)
# color => The color of the bar => (رنگ میله‌ها)
# alpha => The transparency of the bars => (شفافیت میله‌ها)
# density => If True, the histogram is normalized so that the sum of the areas under the graph is equal to 1 => (اگر True باشد، هیستوگرام نرمالیزه می‌شود تا مجموع مساحت زیر نمودار برابر با 1 شود)
# label => Label for histogram used in the guide => (برچسب برای هیستوگرام که در راهنما استفاده می‌شود)

data = np.random.randn(1000)

plt.hist(data, bins=30, color='purple', alpha=0.7, label='Data Distribution')
plt.legend()
plt.show()
```

pie Elements

```python
# x => An array of sizes(آرایه‌ای از اندازه‌ها)
# labels => List of tags for each section(لیست برچسب‌ها برای هر بخش)
# colors => List of colors for each section(لیست رنگ‌ها برای هر بخش)
# explode => List of values ​​to separate segments from the center of the circle( لیست مقادیر برای جدا کردن بخش‌ها از مرکز دایره)
# autopct => Percent formatting (like '%.1f%%')(فرمت‌بندی درصدها (مانند '%.1f%%'))
# shadow => If True, a shadow is added to the chart(اگر True باشد، سایه به نمودار اضافه می‌شود)
# startangle => The starting angle of the chart(زاویه شروع نمودار)

labels = ['A', 'B', 'C', 'D']
sizes = [15, 30, 45, 10]
colors = ['gold', 'yellowgreen', 'lightcoral', 'lightskyblue']
explode = (0, 0.1, 0, 0)  

plt.pie(sizes, explode=explode, labels=labels, colors=colors, autopct='%1.1f%%', shadow=True, startangle=140)
plt.axis('equal')  # Ensuring the chart is circular(اطمینان از دایره‌ای بودن نمودار)
plt.show()
```

boxplot Elements

```python
# x => An array of data or a list of arrays(آرایه‌ای از داده‌ها یا لیستی از آرایه‌ها)
# labels => List of labels for each box( لیست برچسب‌ها برای هر جعبه)
# patch_artist => If True, the boxes are colored( اگر Trueباشد، جعبه‌ها رنگ‌آمیزی می‌شود)
# showmeans => If True, the average of the data is displayed( اگر True باشد، میانگین داده‌ها نمایش داده می‌شود)
# showfliers => If True, outliers are displayed(اگر True باشد، مقادیر پرت (outliers) نمایش داده می‌شوند)

# data(داده)
data = np.random.randn(100, 5)  

# Create a chart(ایجاد نمودار)
plt.figure(figsize=(8, 6))
plt.boxplot(data, labels=['A', 'B', 'C', 'D', 'E'], patch_artist=True)

# Add title and tags( افزودن عنوان و برچسب‌ها)
plt.title('نمودار جعبه‌ای')
plt.xlabel('دسته‌ها')
plt.ylabel('مقادیر')

# Show chart(نمایش نمودار)
plt.show()
```

Customization of graphic elements

```python
# plt.title() => Set chart title(تنظیم عنوان نمودار)
# plt.xlabel() => Set the x-axis label(تنظیم برچسب محور x)
# plt.ylabel() => Set the y-axis label(تنظیم برچسب محور y)
# plt.xlim() و plt.ylim() => Setting axis ranges(تنظیم محدوده‌های محورها)
# plt.xticks() و plt.yticks() => تنظیم مقادیر و برچسب‌های محورها
# plt.grid() => Show grid lines(نمایش خطوط شبکه)
# plt.legend() => Show help(نمایش راهنما)

# data(داده)

x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)

# Create a chart(ایجاد نمودار)
plt.figure(figsize=(10, 6))

# Draw a line graph(رسم نمودار خطی)
plt.plot(x, y1, color='blue', linestyle='-', linewidth=2, label='sin(x)')

# Draw scattered points(رسم نقاط پراکنده)
plt.scatter(x[::10], y2[::10], color='red', marker='o', s=50, label='cos(x)')

# Add title and tags(افزودن عنوان و برچسب‌ها)
plt.title('Sine and Cosine Waves', fontsize=16)
plt.xlabel('X-axis', fontsize=12)
plt.ylabel('Y-axis', fontsize=12)

# Setting axis ranges( تنظیم محدوده‌های محورها)

plt.xlim(0, 10)
plt.ylim(-1.5, 1.5)

# Show grid lines( نمایش خطوط شبکه)
plt.grid(True)

# Show help( نمایش راهنما)
plt.legend(loc='upper right')

#Show chart( نمایش نمودار)
plt.show()
```

Customizing charts(سفارشی‌سازی نمودارها)

```python
# Change colors (تغییر رنگ ها)
plt.plot(x, y, color='red')  # Use the color name(استفاده از نام رنگ)
plt.plot(x, y, color='#FF5733')  # Using hexadecimal code(استفاده از کد هگزادسیمال)

# Changing fonts(تغییر فونت‌ها)
plt.title('Title', fontdict={'fontsize': 20, 'fontweight': 'bold'})
plt.xlabel('X-axis', fontdict={'fontsize': 16})
plt.ylabel('Y-axis', fontdict={'fontsize': 16})

# Changing axis ranges(تغییر محدوده‌های محورها)
plt.xlim(0, 10)  
plt.ylim(-1, 1)  
```

Multiple charts in one figure(چند نمودار در یک Figure)

```python

# data(داده)
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)

# Creating Figures and Axes(ایجاد Figure و Axes)
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(8, 8))  

# Draw graph(رسم نمودار )
ax1.plot(x, y1, color='blue') #Creates two figures in one photo(دو شکل را در یک عکس ایجاد می کند)
ax1.set_title('Sine Wave')
ax1.set_xlabel('X-axis')
ax1.set_ylabel('Y-axis')

ax2.plot(x, y2, color='red')
ax2.set_title('Cosine Wave')
ax2.set_xlabel('X-axis')
ax2.set_ylabel('Y-axis')

# Adjust the spacing between charts(تنظیم فاصله بین نمودارها)
plt.tight_layout()

#Show chart( نمایش نمودار)
plt.show()
plt.savefig('subplots.png')
print("Saved the plot as subplots.png")
```
